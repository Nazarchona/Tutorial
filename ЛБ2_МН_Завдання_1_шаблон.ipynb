```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_csv("/content/CarPrice_Assignment.csv")

print("Розмір датасета:", df.shape)
df.head()
```

 3.	Визначити оптимальну кількість стовпців (видалити зайві, на власний розсуд).

```python
df = df.drop(columns=["car_ID"])

print("Розмір після видалення зайвого стовпця:", df.shape)
print(df.columns)
```

4.	Замініть у таблиці значення "—" на значення NaN. Перевірити наявність пропущених значень. При наявності, замінити пропущені значення на середнє значення.

```python
# Заміна значень "—" на NaN
df = df.replace("—", np.nan)
```

```python
# Перевірка кількості пропущених значень
print("Кількість пропущених значень:")
print(df.isnull().sum())
```

```python
# Заповнення пропущених числових значень середнім
numeric_columns = df.select_dtypes(include=np.number).columns
```

```python
for column in numeric_columns:
    df[column] = df[column].fillna(df[column].mean())

print("\nКількість пропущених значень після заповнення:")
print(df.isnull().sum())
```

6.	Перевірити наявність дублікатів. При наявності видалити дублікати.

```python
# Перевірка на дублікати
duplicates = df.duplicated().sum()

print("Кількість дублікатів:", duplicates)
```

```python
# Видалення дублікатів
df = df.drop_duplicates()

print("Розмір датасету після видалення дублікатів:", df.shape)
```

8. Визначте відхилення (різницю) між показниками IMF (2026) та World Bank (2025) для кожної країни. У яких країнах ці показники найбільше відрізняються (дати відповідь)?

```python
# Середня ціна автомобілів для кожного типу кузова
average_price = df.groupby("carbody")["price"].mean()

# Загальна середня ціна
overall_average = df["price"].mean()

# Відхилення від загальної середньої
price_difference = average_price - overall_average

result = pd.DataFrame({
    "Середня ціна": average_price,
    "Відхилення від загальної середньої": price_difference
})

print("Відхилення середньої ціни за типами кузова:")
display(result.sort_values("Відхилення від загальної середньої", ascending=False))

print("\nТип кузова з найбільшим позитивним відхиленням:")
print(price_difference.idxmax())

print("\nТип кузова з найбільшим негативним відхиленням:")
print(price_difference.idxmin())
```

9.	Обчисліть кореляцію між показниками IMF (2026), World Bank (2025) та United Nations (2024). Які пари змінних мають найвищу кореляцію?

```python
# Вибираємо числові стовпці
numeric_df = df.select_dtypes(include=np.number)

# Кореляційна матриця
correlation_matrix = numeric_df.corr()

print("Кореляційна матриця:")
display(correlation_matrix)
```

```python
# Кореляція всіх числових характеристик з ціною
price_correlation = correlation_matrix["price"].drop("price").sort_values(
    key=abs,
    ascending=False
)

print("Кореляція характеристик із ціною:")
display(price_correlation)
```

Обчислити середнє значення за стовпцями

11.	Обчисліть стандартне відхилення показників для кожної країни. Яка країна має найвищу варіативність у показниках між роками?

```python
# Стандартне відхилення числових характеристик
std_values = df.select_dtypes(include=np.number).std()

print("Стандартне відхилення:")
display(std_values.sort_values(ascending=False).round(2))

# Характеристика з найбільшим стандартним відхиленням
max_std_column = std_values.idxmax()
max_std_value = std_values.max()

print(
    f"\nНайвище стандартне відхилення має "
    f"'{max_std_column}': {max_std_value:.2f}"
)
```

12.	Визначення країни з найвищим та найнижчим показниками: Знайдіть країну з найвищим та найнижчим показниками у кожному з років (IMF (2026), World Bank (2025), United Nations (2024)).

```python
# Найдорожчий автомобіль
max_index = df["price"].idxmax()
max_car = df.loc[max_index]

# Найдешевший автомобіль
min_index = df["price"].idxmin()
min_car = df.loc[min_index]

print("Найдорожчий автомобіль:")
print("Назва:", max_car["CarName"])
print("Ціна:", max_car["price"])

print("\nНайдешевший автомобіль:")
print("Назва:", min_car["CarName"])
print("Ціна:", min_car["price"])
```

13.	Побудуйте гістограму для розподілу показників IMF (2026) серед всіх країн. Який вигляд має розподіл? Чи є країни, що виділяються?

14.	Розрахуйте частку кожної країни в загальному значенні для кожного року (IMF (2026), World Bank (2025), United Nations (2024)).
 Як змінюються частки країн з часом (дати відповідь)?

```python
body_counts = df["carbody"].value_counts()

body_share = (body_counts / len(df)) * 100

result = pd.DataFrame({
    "Кількість автомобілів": body_counts,
    "Частка (%)": body_share.round(2)
})

print("Частка кожного типу кузова:")
display(result)

print(
    "Найбільшу частку в датасеті займають автомобілі типу sedan, "
    "після них hatchback та wagon."
)
```

15.	Візуалізуйте зміни в показниках для кожної країни за три роки на графіку. Які країни показують стабільне зростання або спад (дати відповідь)?

```python
plt.figure(figsize=(10, 6))

plt.scatter(
    df["horsepower"],
    df["price"]
)

plt.xlabel("Потужність двигуна (horsepower)")
plt.ylabel("Ціна автомобіля")
plt.title("Залежність ціни автомобіля від потужності двигуна")

plt.grid(True)

plt.show()

correlation = df["horsepower"].corr(df["price"])

print(
    f"Коефіцієнт кореляції між потужністю двигуна та ціною: "
    f"{correlation:.4f}"
)

print(
    "Спостерігається позитивний зв'язок: зі збільшенням "
    "потужності двигуна ціна автомобіля в середньому зростає."
)
```
